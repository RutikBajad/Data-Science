### *Valid Sudoku*

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
flag = True
for i in s:
    if np.sum(i) != 45:
        print("Not a valid Sudoku.")
        flag = False
        break
for i in s.T:
    if not flag:
        break
    if np.sum(i) != 45:
        print("Not a valid Sudoku.")
        flag = False
        break
if flag == True:
    print("Above sudoku is a valid sudoku")
```

## *General Questions*

```python
import numpy as np
```

```python
data = np.array([
    [18, 85, 78],
    [19, 92, 88],
    [17, 76, 95],
    [18, 65, 70],
    [20, 90, 85]
])
```

```python
# get the shape of the matrix
data.shape
```

```python
# find the average age of students
columns = np.hsplit(data, 3)
print("Average Age :", np.mean(columns[0]))
```

```python
# Extract Maths Marks of all students
print("Maths Marks :\n", columns[1])
```

```python
# find the highest Science Marks
print("Highest marks in science subject is :", np.max(columns[2]))
```

```python
# get details of the student who scored more than 90 in Maths
data[data[:, 1] > 90]
```

```python
# increase Maths marks of all students by 5
data[:, 1] += 5
data
```

```python
# find how many students are younger than 19
len(data[data[:, 0] < 19])
```

```python
# calculate the average marks in each subject (column-wise mean)
print("Maths avg :",np.mean(columns[1]))
print("Science avg :", np.mean(columns[2]))
```

```python
np.mean(data[:, 1:], axis=0)
```

```python
# get data of students who scored at least 80 in both subjects
data[(data[:, 1] >= 80 ) & (data[:, 2] >= 80)]
```

```python
# replace all Science marks < 75 with 0
data[:, 2][data[:, 2] < 75] = 0
```
