### *Numpy Indexing And Slicing Of Vectors*

```python
arr = np.arange(11, 21)
arr
```

```python
arr_slice = arr[3:]
arr_slice
```

```python
arr_slice = arr[:6]
arr_slice
```

```python
arr_slice = arr[3::2]
arr_slice
```

```python
arr[4]
```

### *NUMPY INDEXING AND SLICING OF MATRIX*

```python
arr = np.arange(1, 31).reshape(5, 6)
```

```python
arr
```

```python
slice = arr[0:2, 1:3]
slice
```

```python
slice = arr[2:, 3:5]
slice
```

```python
slice = arr[0]
slice
```

```python
slice = arr[3]
slice
```

```python
slice = arr[:, 2]
slice
```
