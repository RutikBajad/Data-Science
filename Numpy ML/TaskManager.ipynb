```python
tasks = []

def addTask(srno, task_name, status = "Pending"):
    task = {
        "srno" : srno,
        "task_name" : task_name,
        "status" : status,
        "tick" : "[ ✅ ]" if status == "Completed" else "[ ❎ ]"
    }
    tasks.append(task)

def markRead(srno):
    for t in tasks:
        if srno == t["srno"]:
            t["status"] = "Completed"
            t["tick"] = "[ ✅ ]"
            print("Task Marked As Completed.")
            print(f"{t["srno"]}. {t["tick"]} {t["task_name"]} : {t["status"]}.")
        else:
            print("Invalid Task.")

def showTasks():
    for t in tasks:
        print(f"{t["srno"]}. {t["tick"]} {t["task_name"]} : {t["status"]}.")
    print("-"*75)

def deleteTask(srno):
    flag = True
    for t in tasks:
        if srno == t["srno"]:
            tasks.pop(srno-1)
            print("Task Deleted Successfully.")
            flag = False
    if not flag:
        print("Task Not Found.")
    
while True:
    print("<----------Task Manager---------->")
    print("1. Add Task")
    print("2. Mark as Read")
    print("3. Views Tasks")
    print("4. Delete Task")
    print("5. Exit")
    print("<-------------------------------->")
    try :
        choice = int(input("Enter your choice : "))
        if 1 <= choice <= 5:
            match choice:
                case 1:
                    srno = len(tasks) + 1
                    task_name = input("Enter Task : ")
                    addTask(srno, task_name)
                    print("Task Added Successfully.\n")
                case 2:
                    showTasks()
                    srno = int(input("Enter Task no. : "))
                    markRead(srno)
                case 3:
                    print("List of Tasks : ")
                    showTasks()
                case 4:
                    showTasks()
                    srno = int(input("Enter Task no. : "))
                    deleteTask(srno)
                case 5:
                    print("Exiting...")
                    print("Program Closed")
                    break
                case _:
                    print("Invalid choice")
        else:
            print("Enter correct choice!")
    except ValueError as e:
        print(f"Error Occured : {e}")
```
