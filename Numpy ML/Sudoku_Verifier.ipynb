```python
import numpy as np
```

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
valid = True
for row in s:
    if np.sum(row) != 45:
        print(f"Given sudoku is invalid on rows")
        valid = False
        break
```

```python
for column in s.T:
    if np.sum(column) != 45:
        print("Given sudoku is invalid on columns")
        valid = False
        break
```

```python
sum_matrix = []
for i in range(0, 9, 3):
    for j in range(0, 9, 3):
        sum_matrix.append(np.sum(s[i:i+3, j:j+3]))
```

```python
sum_matrix = np.array(sum_matrix).reshape(3,3)
sum_matrix
```

```python
for i in sum_matrix:
    if i.any() != 45:
        valid = False
        break
```

```python
if valid == True:
    print("Sudoku is completely valid!")
else:
    print("Sudoku is invalid!")
```

```python
sum = []
sum.append(np.sum(s[:3,:3]))
sum.append(np.sum(s[3:6, :3]))
sum.append(np.sum(s[6:, :3]))
sum.append(np.sum(s[:3,3:6]))
sum.append(np.sum(s[3:6, 3:6]))
sum.append(np.sum(s[6:, 3:6]))
sum.append(np.sum(s[:3,6:]))
sum.append(np.sum(s[3:6, 6:]))
sum.append(np.sum(s[6:, 6:]))
```

```python
for e in sum:
    if e != 45:
        valid = False
```
