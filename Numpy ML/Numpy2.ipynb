# *Array Operations*

### *Arithmetic Operations*

```python
a1 = np.arange(1, 6)
a2 = np.arange(6, 11)
print(a1, a2)
```

```python
add = a1 + a2
add
```

```python
sub = a1 - a2
sub
```

```python
mul = a1 * a2
mul
```

```python
power = a1 ** a2
power
```

```python
div = a1 / a2
div
```

```python
fldiv = a1 // a2
fldiv
```

### *Broadcasting*

```python
l = [10, 20, 30, 40]
arr = np.array(l)
```

```python
for i in range(len(l)):
    l[i] += 10
l
```

```python
arr += 10 #broadcasting adding 10 to each item of array
arr
```

```python
a = np.arange(1, 26).reshape(5, 5)
a
```

```python
a += 10
a 
```

```python
a * 2
```

### *Deep and shallow copy*

```python
a = np.arange(1, 10)
a
```

```python
slice  = a[:5]
```

```python
slice = slice * 10
slice

# slice *= 10 will directly affect original a
```

```python
slice *= 10
slice
```

```python
a
```

```python
a[4:6] * 10
```

```python
a
```

```python
b = a
```

```python
b[0] = 99
```

```python
b
```

```python
a
```

```python
b = a.copy() #shallow copy
```

```python
b
```

```python
b[0] = 1
```

```python
b
```

```python
a
```

```python
a = [[1, 2],[3, 4]]
b = a.copy()
print(a)
print(b)
```

```python
b[0].append(99)
print(a)
print(b)
```

```python
import copy
m = [[1, 2],[3, 4]]
n = copy.deepcopy(m)
print(m)
print(n)
```

```python
n[0].append(99)
print(m)
print(n)
```

## *Matrix Operations*

```python
A = np.array([[1, 2],[3, 4]])
B = np.array([[5, 6],[7, 8]])
```

```python
A
```

```python
B
```

```python
prod = A*B
prod # each element is getting multiplicated to the respective element, not actually a matrix multiplication
```

```python
prod = A @ B 
prod
```

```python
prod = np.dot(A, B)
prod
```

### *Splitting array*

```python
c = np.arange(16).reshape(4, 4)
c
```

```python
np.hsplit(c, 2)
```
