```python
for i in range(1, 101):
    n = str(i)
    if n[::-1] == str(i):
        print(n, end="\t")
```

```python
n = 1
while n <= 5000000:
    i = n
    num = 0
    while i > 0:
        ld = i%10
        i //= 10
        num = num * 10 + ld
    if num == n:
        print(n, end="\t")
    n += 1
```
