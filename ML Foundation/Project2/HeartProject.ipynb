# Data Preprocessing and Cleaning

```python
df_encode = pd.get_dummies(df, drop_first=True)
```

```python
df_encode
```

```python
df_encode = df_encode.astype(int)
```

```python
df_encode
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
numeric_col = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']
df_encode[numeric_col] = scaler.fit_transform(df_encode[numeric_col])
```

```python
df_encode
```

```python
df_encode.columns
```

```python
X = df_encode.drop(columns=['HeartDisease'])
y = df_encode['HeartDisease']
```

```python
from scipy.stats import pearsonr
import pandas as pd

results = []

for feature in X.columns:
    r, p = pearsonr(X[feature], y)

    results.append({
        'Feature': feature,
        'Correlation': r,
        'Abs_Correlation': abs(r),
        'P-value': p
    })

corr_df = pd.DataFrame(results)

corr_df = corr_df.sort_values(
    by='Abs_Correlation',
    ascending=False
)

corr_df
```

```python
cat_features = [
    'Sex_M',
    'ChestPainType_ATA',
    'ChestPainType_NAP',
    'ChestPainType_TA',
    'RestingECG_Normal',
    'RestingECG_ST',
    'ExerciseAngina_Y',
    'ST_Slope_Flat',
    'ST_Slope_Up'
]
```

```python
from scipy.stats import chi2_contingency

alpha = 0.05

chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(df_encode[col], df_encode['HeartDisease'])
    
    chi2_stats, p_val, _, _ = chi2_contingency(contingency)
    
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    
    chi2_results[col] = {
        'chi2_statistics': chi2_stats,
        'p_value': p_val,
        'Decision': decision
    }
chi2_results_df = pd.DataFrame(chi2_results).T

chi2_results_df
```
