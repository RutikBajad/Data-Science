```python
import numpy as np
import pandas as pd
```

## *Finding Missing Data*

```python
data = {
    'A' : [1, 2, np.nan, 4, 5],
    'B' : [np.nan, 2, 3, 4, 5],
    'C' : [1, 2, 3, np.nan, np.nan],
    'D' : [1, np.nan, np.nan, np.nan, 5]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any() # checks where if any column has null values
```

## *Removing Missing Data*

```python
df
```

```python
df.dropna() # removes rows that has null values
# use inplace = True to actually delete data
```

```python
df
```

```python
df.dropna(thresh=3) # removes rows having less than three non-null values
```

## *Filling the missing data*

```python
df
```

```python
df.fillna(0)
```

```python
values = {'A':0, 'B':100, 'C':300, 'D':400}
df.fillna(value=values)
```

```python
df
```

```python
df.fillna(df.mean())
```

```python
df.mean()
```
