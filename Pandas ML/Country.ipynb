####  *Which country has the highest population*

```python
df[df['population'] == df['population'].max()][['country', 'population']]
```

#### *What is the capital of the country with highest population*

```python
df[df['population'] == df['population'].max()][['country','capital_city']]
```

#### *Which country has the least population*

```python
df[df['population'] == df['population'].min()][['country', 'population']]
```

#### *What is the capital of the country with least population* 

```python
df[df['population'] == df['population'].min()][['country','capital_city']]
```

#### *give me 5 countries with highest democratic score*

```python
df_sorted_by_democratic_score = df.sort_values(by='democracy_score', ascending=False)
df_sorted_by_democratic_score[['country', 'democracy_score']].head()
```

#### *how many total regions are there*

```python
count = 0
for i in df.groupby('region'):
    count += 1
print(f"There are total {count} regions.")
```

```python
print(df['region'].value_counts().count())
```

#### *who is the political leader of the 2nd highest populated country*

```python
df_sorted_by_population = df.sort_values(by='population', ascending=False)
df_sorted_by_population.iloc[1][['country','political_leader']]
```

```python
df[df['population'] == df['population'].nlargest(2).iloc[1]][['country','political_leader']]
```

#### *how many countries are there whose political leaders are unknown*

```python
df[df['political_leader'].isna()]['country'].count()
```

#### *how many countries have republic in their name*

```python
count = 0
def countries_having_republic(txt):
    global count
    if "republic" in txt.lower():
        count += 1
    return txt
df['country_long'] = df['country_long'].apply(countries_having_republic)
print(count)
```
