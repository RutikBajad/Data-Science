## *DataFrames*

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name' : ["John", "Anna", "Peter", "Linda"],
    "Age" : [28, 34, 29, 42],
    "City" : ["New York", "Paris", "Berlin", "London"],
    "Salary" : [65000, 70000, 62000, 85000]
}
```

#### *Creating a DataFrame*

```python
df = pd.DataFrame(data) #Multiple series
```

```python
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
```

```python
df2 = pd.DataFrame(data_list, columns=['Name', 'Age', 'City', 'Salary'])
df2
```

#### *Selection and Indexing of Columns*

```python
df2["Name"]
```

```python
df2[["Name", "Age"]]
```

#### *Creating a new column*

```python
df2["Designation"] = ["Doctor", "Engineer", "Astraunaut", "Plumber"]
df2
```

#### *Removing a column*

```python
df2.drop("Designation", axis=1) 
# it only shows the snapshot of how data looks after deletion
```

```python
df2
```

```python
df2.drop(["Designation", "Salary"], axis=1, inplace=True) 
# it actually deletes the columns
```

```python
df2
```

```python
# i restored df here by rerunning the data
```

```python
df2.drop(1) #inplace will actually delete data
```

```python
df2.loc[0]
```

```python
df2.loc[[0, 2]]
```

#### *Conditional Selection*

```python
df2
```

```python
# Want to see those people whose age is above 30
df2[df2['Age'] > 30]
```

```python
# Want to see those people whose age is above 30 and are from Paris
df2[(df2['Age'] > 30) & (df2['City'] == 'Paris')]
```
