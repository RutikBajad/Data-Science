## *Create a column for no of episodes*

```python
df.loc[1]
```

```python
df.loc[2]['Title']
```

```python
def extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ')':
            check = False
            return data
        if check:
            data += i 
        if i == '(':
            check = True   
```

```python
df['Episodes'] = df['Title'].apply(extract_episodes)
```

```python
df.head()
```

```python
df['Episodes'] = df['Episodes'].str.replace(" eps", "")
```

```python
df.head()
```

```python
df['Episodes'] = df['Episodes'].apply(lambda x: int(x))
```

## *Make a new column for time stamp*

```python
df['Title'].loc[10]
```

```python
def extract_timestamp(txt):
    check = False
    timestamp = ""
    k = 0
    for i in txt:
        if k == 19:
            return timestamp
        if check and k < 19:
            timestamp += i
            k += 1
        if i == ')':
            check = True
        
```

```python
df['TimeStamp'] = df['Title'].apply(extract_timestamp)
```

```python
df.head()
```

```python
from dateutil.relativedelta import relativedelta
from datetime import datetime

def calculate_total_months(period):
    try:
        start_str, end_str = period.split(' - ')
        start_date = datetime.strptime(start_str, '%b %Y')
        end_date = datetime.strptime(end_str, '%b %Y')
        r = relativedelta(end_date, start_date)
        return r.years * 12 + r.months + 1 # +1 to include the starting month
    except:
        return None
```

```python
df['Months'] = df['TimeStamp'].apply(calculate_total_months)
```
