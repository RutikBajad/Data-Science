```python
import seaborn as sns
import matplotlib.pyplot as plt
```


We use distribution plots to visualize the distribution of quantitative data.

Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

- distplot/histplot
- jointplot
- pairplot
- rugplot

```python
df = sns.load_dataset('tips') # datasets of official seaborn, requires internet to load
df
```

```python
df['size'].unique()
```

```python
sns.displot(df['total_bill'])
```

```python
sns.displot(df['total_bill'], bins = 5)
```

```python
sns.histplot(df['total_bill'], bins=15, kde=True) # same as displot
```

```python
plt.subplot(1, 2, 1)
sns.histplot(df['total_bill'], bins=15, kde=True)
plt.subplot(1, 2, 2)
sns.histplot(df['tip'], bins=20)
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind='scatter')
# kind = scatter default
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df, kind='reg')
```

```python
sns.pairplot(df, hue='sex', palette='rainbow')
```

```python
sns.pairplot(df, hue='size', palette='rainbow')
```

```python
sns.rugplot(df['total_bill'])
```
