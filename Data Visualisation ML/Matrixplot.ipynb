```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights = sns.load_dataset('flights')
```

```python
flights
```

```python
tips = sns.load_dataset('tips')
tips
```

## *Heat Map*

```python
tipscorr = tips[['total_bill', 'tip', 'size']]
tipscorr
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr())
```

```python
sns.heatmap(tipscorr.corr(), annot=True)
```

## *Cluster Map*

```python
sns.clustermap(tipscorr.corr())
```

## *pivot table Heat Map*

```python
pvt_flights = flights.pivot_table(values = 'passengers', index = 'month', columns = 'year')
pvt_flights
```

```python
sns.heatmap(pv_flights)
```
