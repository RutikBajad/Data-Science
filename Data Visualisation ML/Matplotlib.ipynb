```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#### *Matplotlib Basics*

```python
x = np.linspace(0, 5, 11)
y = x ** 2
print(x)
print(y)
```

```python
plt.plot(x, y)
```

```python
plt.plot(x, y)
plt.title("Hello")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

#### *Subplots in Matplotlib*

```python
plt.subplot(2, 2, 1)
plt.plot(y, x)
plt.subplot(2, 2, 4)
plt.plot(x, y)
plt.subplot(2, 2, 3)
plt.plot(x**2, y)
plt.subplot(2, 2, 2)
plt.plot(x**3, y)
```

#### *Object Oriented way*

```python
fig = plt.figure()
fig.add_axes([0.1, 0.1, 0.8, 0.8])
```

```python
fig = plt.figure()
fig.add_axes([0.1, 0.1, 0.6, 0.9])
```

##### *Matplotlib allows the aspect ratio, DPI and figure size to be specified when the figure object is created. You can use the figsize and dpi keyword arguments.*

##### *figsize is a tuple of the width and height of the figure in inches. dpi is the dots-per-inch (pixel per inch)*

```python
fig = plt.figure(dpi=800)  # dpi is for clarity
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x, y)
axis2 = fig.add_axes([0.3, 0.5, 0.3, 0.3])
axis2.plot(x, y)
```

## *Saving your plot*

```python
fig = plt.figure(figsize=(5, 5), dpi=300)
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax.plot(x, y)
plt.savefig("basicplot.png")
```
