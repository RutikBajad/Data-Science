### A tech company claims that the average battery life of their new phone is 30 hours (mu = 30) with a known population standard deviation of 4 hours (sigma = 4).
### A consumer group tests a random sample of 36 phones (n = 36) and finds an average battery life of 28.5 hours (x(bar) = 28.5).

```python
## Null Hypothesis (Ho) = Battery life was correct
## Alternate Hypothesis (H1) = Battery life was false
```

```python
import numpy as np
from scipy.stats import norm
```

```python
# Sample data representing battery life (in hours) for 36 phones
sample_data = np.array([
    26.2, 29.1, 31.4, 27.5, 25.8, 30.2, 28.9, 32.1, 24.5, 29.8,
    27.1, 30.6, 28.4, 26.9, 33.2, 25.1, 29.4, 27.8, 31.1, 28.3,
    26.5, 30.1, 29.0, 27.4, 32.5, 24.9, 28.7, 30.9, 26.1, 29.3,
    27.6, 31.8, 25.5, 28.1, 30.4, 26.9
])
population_mean = 30
population_std = 4
sample_mean = np.mean(sample_data)
n = len(sample_data)
```

```python
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(n))
z_score
```

```python
p_value = 2 * ( 1 - norm.cdf(abs(z_score)))
p_value
```

```python
alpha = 0.05
```

```python
if alpha > p_value:
    print('Null Hypothesis Rejected')
else:
    print('Null Hypothesis was correct')
```
