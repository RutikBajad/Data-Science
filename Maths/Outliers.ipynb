```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
arr = np.array([2, 3, 4, 6, 7, 8, 9, 12, 13, 16, 17, 23, 25, 27, 34, 37, 201])
arr
```

```python
Q1 = np.percentile(arr, 25)
Q1
```

```python
Q2 = np.percentile(arr, 50)
Q2
```

```python
Q3 = np.percentile(arr, 75)
Q3
```

```python
IQR = Q3 - Q1
IQR
```

```python
Lower_fence = Q1 - 1.5 * IQR
Lower_fence
```

```python
Upper_fence = Q3 + 1.5 * IQR
Upper_fence
```

```python
plt.figure(figsize=[7, 5])
plt.boxplot(arr)
plt.title('Box Plot Visualisation to detect outliers')
plt.show()
```

```python
l = []
for i in arr:
    if i >= Lower_fence and i <= Upper_fence:
        l.append(i)
arr2 = np.array(l)
print(arr)
print(arr2)
```

```python
sns.boxplot(x = arr)
```

```python
sns.boxplot(x = arr2)
```
