```python
data.drop(data.index[17], inplace=True)
```

```python
data['Age'] = data['Age'].fillna(int(data['Age'].mean()))

data['City'] = data['City'].fillna(data['City'].mode()[0])

data['Salary'] = data['Salary'].fillna(int(data['Salary'].mean()))
```

```python
data.loc[20, 'Gender'] = 'Male'
```

```python
data.loc[13, 'Email'] = 'rahul@example.com'
```

```python
data.info()
```

```python
data.drop(14, inplace=True)
```

```python
data['Joining_Date'] = pd.to_datetime(data['Joining_Date'], format='mixed')
```

```python
data['Email'] = data['Email'].str.lower()
data['City'] = data['City'].str.lower()
data['Gender'] = data['Gender'].str.lower()
data['Name'] = data['Name'].str.lower()
```

```python
data.loc[(data['Age'] < 0) | (data['Age'] > 100), 'Age'] = data['Age'].mean().astype(int)
```

```python
data.loc[(data['Salary'] < 0), 'Salary'] = data['Salary'].mean().astype(int)
```

```python
data['City'] = data['City'].str.strip()
```

```python
for x in data.index:
    if data.loc[x, 'Gender'] == 'female':
        data.loc[x, 'Gender'] = 'f'
    if data.loc[x, 'Gender'] == 'male':
        data.loc[x, 'Gender'] = 'm'
```

```python
data.loc[9, 'Email'] = 'sneha@example.com'
```

```python
data['ID'] = data['ID'].astype(int)
```

```python
data.drop_duplicates(inplace=True)
```
